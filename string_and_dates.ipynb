```python
emp=spark.createDataFrame(data=emp_data,schema=emp_schema)
display(emp)
```

```python
from pyspark.sql.functions import when, col

emp_gender_categorization=emp.select("employee_id","department_id","name","age","gender","salary","hire_date", 
    when(col("gender") == "Female", "F")
    .when(col("gender") == "Male", "M")
    .otherwise("Unknown")
    .alias("genderStatus"))

```

```python
emp_gender_categorization=emp.withColumn("GenderStatus", when(col("gender")=="Male","M")
                                                              .when(col("gender")=="Female","F")
                                                              .otherwise("Unknown"))\
                                .withColumn("Seniority", when(col("hire_date").between("2000-01-01","2014-01-01"),"Senior")
                                        .when(col("hire_date").between("2015","2019"),"Mid-Senior")
                                        .otherwise("Junior"))


                                                              
```

```python
emp_gender_categorization.show()
```

```python
emp_gender_categorization.printSchema()
```

```python

from pyspark.sql.functions import col, expr

emp_gender_categorization=emp_gender_categorization.selectExpr("employee_id","department_id", "name", "cast(age as int) as age","gender","salary","hire_date","GenderStatus","Seniority")
emp_gender_categorization.printSchema()
```

```python
age_categorization=emp_gender_categorization.withColumn("AgeCategory", expr("case when age between 0 and 18 then 'Child' when age between 19 and 30 then 'Young' when age between 31 and 45 then 'Adult' else 'Aged' end"))
age_categorization.show()
```
